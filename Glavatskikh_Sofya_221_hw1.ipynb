**Пункт 1**

```python
import os  # importing needed libraries (excuse me for english comments, it was tiresome for me to change between layouts within the code)
import pymystem3 as mystem
from string import punctuation

with open("invitation.txt", encoding="utf-8") as rawtext:  # opening the original file = "Invitation to a Beheading" by Nabokov
    text = rawtext.read()
    paragraphs = " ".join(text.split("/n"))
    words = ' '.join([w.strip(punctuation) for w in paragraphs.split()])  # decided to clear text from punctuation for punctuation points
```

**Пункт 2**

```python
%%time
lemmatizer = mystem.Mystem()
lemmaslist = lemmatizer.lemmatize(words)  # lemmatizing the text
lemmaslist
# for some reason the list of lemmas isn't showing fully in jupyter notebook so I'm not sure if the time is correctly measured :(
# even though the list of lemmas is full when written in the file
```

```python
with open("invitation_lemmatized.txt", "w", encoding="utf-8") as newfile:  # saving joined (from list of lemmas) string
    newfile.write("".join(lemmaslist))
```

```python
%%time
import jsonlines
from pymorphy3 import MorphAnalyzer
analyzer = MorphAnalyzer()  # creating analyzer for tokens from the text
save = []  # for comfortable iterating in following punkts
with jsonlines.open("pymorphyanalysis.jsonl", "w") as writer:  # opening jsonlines-file
# let's create and fill each new dictionary within jsonlines-file with each cycle iteration
    for x in nltktokens:
        dict = {}
        element = analyzer.parse(x)[0]  # analyzing a token for this iteration
        dict["lemma"] = element.normal_form
        dict["word"] = element.word
        dict["pos"] = element.tag.POS
        save.append(dict)
        writer.write(dict)  # writing analysed elements into jsonlines
```

**Пункт 4**

Для начала -- считаем долю частей речи, здесь два способа. Вначале самый простой, но немного грубый способ проверить доли всех частей речи, когда тэги есть в готовом массиве.

```python
for pos in ['NOUN', 'ADJF', 'ADJS', 'COMP', 'VERB', 'INFN', 'PRTF', 'PRTS', 'GRND', 'NUMR', 'ADVB', 'NPRO', 'PRED', 'PREP', 'CONJ', 'PRCL', 'INTJ']:
    c = 0
    for x in save:  # checking in safe list, that we created with jsonlines-file (for some reason jsonlines-file wasn't comfortably iterable like that)
        if x["pos"] == pos:
            c += 1  # counting number of this exact p.o.s.
    print("Доля", pos, "в тексте равна ", c / len(save))  # dividing by the whole number of lines in the file
```

А это второй способ, который мне показался логичнее для случаев, когда нам не известен полный список тэгов (сам цикл считает уникальные тэги через Counter), но почему-то значения долей в нем несколько отличаются, и даже при помощи функции для проверки на кириллицу не удалось избавиться от случаев None в общем подсчёте...

```python
import re
from collections import Counter

def is_cyrillic(s):  # to check if the word is in cyrillic
    lower = set('абвгдеёжзийклмнопрстуфхцчшщъыьэюя')
    return lower.intersection(s.lower()) == set(s)

counter = 0  # for counting all the lines in the jsonlines file                         
countdict = Counter()  # here we'll put each new p.o.s. tag that the cycle stumbles upon as a new key
with jsonlines.open("pymorphyanalysis.jsonl") as reader:
    for x in reader:
        if is_cyrillic(x["word"]):  # if the word is cyryllic ( i've tried :( )
            countdict[x["pos"]] += 1
print(countdict)
for x in countdict:
    dolya = countdict[x] / len(save)  # 
    print(f"Доля {x} в тексте равна {dolya}")
```

Дальше о самых частотных глаголах и наречиях -- это тоже можно сделать серез словарь Counter.

```python
countverb = Counter()
countadvb = Counter()
verbs = []
adverbs = []
for x in save:
    if x["pos"] == "VERB":
        verbs.append(x["lemma"])
        countverb[x["lemma"]] += 1
    if x["pos"] == "ADVB":
        adverbs.append(x["lemma"])
        countadvb[x["lemma"]] += 1
print("Самые частотные глаголы:")
for x in countverb.most_common(20):
    print(x[0], x[1])
print("\nСамые частотные наречия:")
for x in countadvb.most_common(20):
    print(x[0], x[1])
```

**Пункт 5**

```python
from nltk.util import bigrams, trigrams
lemmasclear = []  # we need to remove whitespaces and the rest of excessive symbols from the list of lemmas from Пункт 1
for x in lemmaslist:
    if x.isalpha():
        lemmasclear.append(x)
bigramslist = list(bigrams(lemmasclear))  # creating lists of all bigrams and trigrams in the text, using cleared lemmas
trigramslist = list(trigrams(lemmasclear))
countbi = Counter()  # creating counters for bigrams and trigrams
counttri = Counter()
for x in bigramslist:  # counting
    countbi[x] += 1
for y in trigramslist:
    counttri[y] += 1
print("Самые частотные биграммы:")  # output
for x in countbi.most_common(25):
    print(" ".join(x[0]))
print("\nCамые частотные триграммы:")
for y in counttri.most_common(25):
    print(" ".join(y[0]))
```
